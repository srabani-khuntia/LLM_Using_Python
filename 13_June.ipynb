# **Pandas**







```python
import pandas as pd

# Sample DataFrames for demonstration
data1 = {'common_column': ['A', 'B', 'C'], 'value1': [10, 20, 30]}
df1 = pd.DataFrame(data1)

data2 = {'common_column': ['B', 'C', 'D'], 'value2': [200, 300, 400]}
df2 = pd.DataFrame(data2)

data_pivot = {'date': ['2023-01-01', '2023-01-01', '2023-01-02', '2023-01-02'],
              'category': ['X', 'Y', 'X', 'Y'],
              'value': [1, 2, 3, 4]}
df = pd.DataFrame(data_pivot)

#Merging two dataframes on a common column
merged_df = pd.merge(df1, df2, on='common_column') # without common column, you can't arbitrarily merge

#Concatenating two dataframes
concatenated_df = pd.concat([df1, df2], axis=0)

#Reshaping a dataframe with pivot
pivot_df = df.pivot(index='date', columns='category', values='value')

#Stacking a dataframe
stacked_df = df.stack()

#Unstacking a dataframe
unstacked_df = df.unstack()

#Display the trasformed dataframes
print(merged_df)
print(concatenated_df)
print(pivot_df)
print(stacked_df)
print(unstacked_df)
```

## **Handling categorical data**

```python
#Load dataset
data = {'category_column': ['apple', 'banana', 'apple', 'orange', 'banana', 'apple'],
        'value_column': [10, 20, 15, 25, 30, 12]}
df = pd.DataFrame(data)

#Convert a column to categorical data
df['category_column'] = df['category_column'].astype('category')

#Display the df with categorical data
print(df.head())
```

## **One-Hot Encoding**

> transforms categorical data suitable for ML algorithms by creating binary columns for each category.



```python
#One hot encode the categorical column
df_one_hot_encoded = pd.get_dummies(df, columns=['category_column'])

#Display the df with one-hot encoded data
print(df_one_hot_encoded.head())
```

## **Label Encoding**

```python
from sklearn.preprocessing import LabelEncoder

#Initialize the label encoder
label_encoder = LabelEncoder()

#Apply label encoding to the categorical column
df['encoded_column'] = label_encoder.fit_transform(df['category_column'])

#Display the dataframe with encoded categorical data
print(df.head())
```

## **Groupby**

> allows to split a df into groups based on one or more columns. Apply a function to each group and combines the result.





```python
#Sample dataset
data = {'category': ['A', 'A', 'B', 'B','C', 'C'],
        'sales': [100, 150, 200, 130, 90, 120]}
df = pd.DataFrame(data)

#Group by category and sum sales
grouped = df.groupby('category')['sales'].sum()

#Display the grouped data
print(grouped)
```

## **Pivot Table**

```python
#Sample data
data = {'category': ['A', 'A', 'B', 'B', 'C', 'C'],
        'year': ['2020', '2021', '2020', '2021', '2020', '2021'],
        'sales': [100, 150, 200, 130, 90, 120]}
df = pd.DataFrame(data)

#Create a pivot table
pivot_table = df.pivot_table(index='category', columns='year', values='sales', aggfunc='sum')

#Display the pivot table
print(pivot_table)
```

## **Indexing**

```python
#Sample dataset with datetime index
data = {'A':[1,2,-3,4],
        'B':[4,-5,6,-7],
        'C':[7,-8,9,10]}

df = pd.DataFrame(data, index=pd.date_range('2020-01-01', periods=4))

#Selecting rows by label
row = df.loc['2020-01-01']

#Selecting columns by label range
cols = df.loc[:, 'A':'C']

#Boolean indexing with labels
filtered = df.loc[df['A']>0]

print(row)
print(cols)
print(filtered)
```

### **iloc**

> position based indexing is crucial when data's structure is more important than the labels.



```python
#Selecting rows by position
row = df.iloc[0]

#Selecting columns by position range
cols = df.iloc[:,1:3]

#Slicing rows and columns by position
slice = df.iloc[1:5, 0:2]

print(row)
print(cols)
print(slice)
```

### **Multi indexing**

```python
from sklearn.base import np
#Creting and manipulating Multiindex objects
arrays = [np.array(['bar', 'bar', 'baz', 'baz']), np.array(['one', 'two', 'one', 'two'])]

df_multi = pd.DataFrame(np.random.randn(4,2), index=arrays, columns=['A', 'B'])
df_multi.index.names = ['first', 'second']

print(df_multi)
```

## **apply()**

> to be used judiciously. Helpful in applying custom functions row-wise or column-wise, it can be slower than vectorized operations.



```python
def custom_function(x):
    return x * 2

#Using apply with a lambda function
df['processed_column'] = df['A'].apply(lambda x: custom_function(x))

#Instead consider using vectorised operations:
df['processed_column'] = custom_function(df['A'])

#Check memory usage of each column
print(df.memory_usage())

#Optimize memory usage by converting data types
df['A'] = pd.to_numeric(df['A'], downcast='integer')

#Check memory usage after optimization
print(df.memory_usage())
```
