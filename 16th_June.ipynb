**Line plots, vizualizing trends and progress:** to see how a variable changes wrt another.


**Applications:** showing trends over time, epochs during training etc.

```python
import matplotlib.pyplot as plt
import numpy as np

#Sample data: Simulating model accuracy over epochs
epochs = np.arange(1,11)
accuracy = np.linspace(0.5, 0.95, 10)

plt.plot(epochs, accuracy) #this is for a line plot, for a bar plot: plt.bar, similarly for others
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.title('Model Accuracy During training')
plt.grid(True) #Add gridlines

```

**Scatter Plot**

```python
#Sample data: Simulating the relationship between learning rate and model loss
learning_rates = np.array([0.001, 0.01, 0.1, 1])
loss_values = np.array([0.8, 0.5, 0.2, 0.3])

plt.scatter(learning_rates, loss_values, marker='o', s=100) #o is circle, s is size
plt.xlabel('Learning Rate')
plt.ylabel('Loss')
plt.title("Effect of Learning rate On model loss")
plt.xscale('log') #Use logarithmic scale for learning rate
plt.show()
```

**Bar Graph**

```python
#Sample data: Comparing the accuracy of different ML models
models = ['Logistic Regression', 'Decision Tree', 'SVM', 'Random Forest']
accuracies = [0.75, 0.82, 0.85, 0.90]

plt.bar(models, accuracies, color=['blue', 'green', 'red', 'purple']) #Lighter color preferred over vibrant for better UI and visibility
plt.xlabel('ML Models')
plt.ylabel('Accuracy')
plt.title('Model Accuracy Comparison')
plt.ylim(0,1) #Set y-axis limits for better visualization
plt.show()
```

**Histogram**

```python
#Sample data: generating data with normal distribution
data = np.random.randn(1000)

plt.hist(data, bins=30, edgecolor='black')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Histogram of Randomly Generated Data')
plt.show()
```

### **Confusion Matrix**

The level of confusion needs to be hard coded.

```python
from itertools import product

#sample confusion matrix(can be replaced with the model's output)
confusion_matrix = np.array([[90,10],
                             [5,95]])

#Function to plot the confusion matrix with labels
def plot_confusion_matrix(cm, classes, title='Confusion Matrix', cmap=plt.cm.Blues):
  plt.imshow(cm, interpolation='nearest', cmap=cmap)
  plt.title(title)
  plt.colorbar()
  tick_marks = np.arange(len(classes))
  plt.xticks(tick_marks, classes, rotation=45)
  plt.yticks(tick_marks, classes)

  thresh = cm.max() / 2. #divides the cm to 2 segments. Can be 4 as well
  for i, j in product(range(cm.shape[0]), range(cm.shape[1])):
    plt.text(j, i, format(cm[i, j], 'd'), horizontalalignment='center', color='white' if cm[i, j] > thresh else 'black')

  plt.tight_layout()
  plt.ylabel('True Label')
  plt.xlabel('Predicted Label')

#Example class labels
class_names = ['Class 0', 'Class 1']

#Plot the confusion matrix
plt.figure()
plot_confusion_matrix(confusion_matrix, classes=class_names)
plt.show()
```

### **Visualize an Image and feature maps with CNN**

```python
image = np.random.randn(28,28) #Simulating a 28x28 grayscale image

#Sample feature maps(to be replaced with the output of the CNN layer)
feature_maps = np.random.randn(8,10,10) #8 filters each producing a 10x10 feature map

#Plot the original image
plt.figure()
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

#Plot the feature maps
plt.figure(figsize=(10,5)) #adjust figure size as needed
for i in range(feature_maps.shape[0]):
  plt.subplot(2,4,i+1) #Create subplots(2 rows, 4 columns)
  plt.imshow(feature_maps[i], cmap='viridis')
  plt.title(f'Feature Map {i+1}')
  plt.axis('off')

plt.tight_layout()
```
